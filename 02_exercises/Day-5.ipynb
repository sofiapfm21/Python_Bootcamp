## Pandas

```python
import pandas as pd
```

#### 1. Create a pandas DataFrame named `df` using the following table of names, ages, and scores.

| Name  | Age | Score |
|-------|-----|-------|
| Anna  | 23  | 88    |
| Ben   | 35  | 92    |
| Cara  | 29  | 79    |
| David | 41  | 85    |

```python
data = {'Name': ['Anna', 'Ben', 'Cara', 'David'],
        'Age': [23, 35, 29, 41],
        'Score': [88, 92, 79, 85]}
df = pd.DataFrame(data)
display(df)
```

#### 2. Use pandas methods on `df` to find the dataset’s dimensions (number of rows and columns) and calculate the average score.

```python
dimensions = df.shape
print(f"Dataset Dimensions (rows, columns): {dimensions}")

average_score = df['Score'].mean()
print(f"Average Score: {average_score}")
```

#### 3. Filter the DataFrame `df` to keep only rows where the score is greater than or equal to 85 **and** the age is greater than 30.


```python
filtered_df = df[(df['Score'] >= 85) & (df['Age'] > 30)]
display(filtered_df)
```

#### 4. Create a new column `Score_decimal` by dividing the `Score` column by 100.

```python
# Your code goes here
df['Score_decimal'] = df['Score'] / 100
display(df)
```

## Matplotlib

```python
import matplotlib.pyplot as plt
```

#### 1. **Bar Plot.** Create a bar chart showing each person’s Name on the x-axis and their Score on the y-axis.

```python
plt.figure(figsize=(10, 6))
plt.bar(df['Name'], df['Score'], color='skyblue')
plt.xlabel('Name')
plt.ylabel('Score')
plt.title('Scores by Name')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#### 2. **Scatter Plot.** Create a scatter plot for df showing the relationship between `Age ` and `Score`.

```python
plt.figure(figsize=(10, 6))
plt.scatter(df['Age'], df['Score'], color='red', alpha=0.7)
plt.xlabel('Age')
plt.ylabel('Score')
plt.title('Age vs. Score Scatter Plot')
plt.grid(True)
plt.show()
```

#### 3. **Histogram**. Create a histogram to show the distribution of the `Score` column.

#### 4. **Box Plot**. Create a box plot to visualize the distribution of the `Age` column.

```python
plt.figure(figsize=(8, 6))
plt.boxplot(df['Age'])
plt.xlabel('Age')
plt.title('Distribution of Age')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['Score'], bins=5, color='lightgreen', edgecolor='black')
plt.xlabel('Score')
plt.ylabel('Frequency')
plt.title('Distribution of Scores')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
