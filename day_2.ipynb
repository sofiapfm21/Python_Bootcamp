#### 2. Write a program to accept two integer numbers from the user and calculate their product (multiplication).

````python
num1 = int(input("Enter the first integer: "))
num2 = int(input("Enter the second integer: "))

product = num1 * num2

print(f"The product of {num1} and {num2} is: {product}")
````

To create a program that accepts two integer numbers from the user and calculates their product, we follow these logical steps:

1.  **Get First Integer Input:**
    *   We use the `input()` function to display a prompt to the user, asking them to "Enter the first integer:".
    *   The `input()` function reads whatever the user types as a string.
    *   Since we need to perform mathematical operations, we convert this string to an integer using the `int()` function and store it in a variable called `num1`.

2.  **Get Second Integer Input:**
    *   Similarly, we use `input()` to ask for the "second integer".
    *   Again, we convert the user's input from a string to an integer using `int()` and store it in `num2`.

3.  **Calculate Product:**
    *   Once we have both numbers as integers (`num1` and `num2`), we perform the multiplication operation using the `*` operator.
    *   The result of this multiplication is stored in a new variable called `product`.

4.  **Display Result:**
    *   Finally, we use the `print()` function to show the result to the user.
    *   An f-string (`f"The product of {num1} and {num2} is: {product}"`) is used for clear and readable output, embedding the values of `num1`, `num2`, and `product` directly into the message.

This sequence ensures that the program correctly obtains numerical input, performs the required calculation, and presents the answer in an understandable format.
