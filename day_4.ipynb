#### 1. Create a class called `Book` that has the attributes title, author, pages, price. It also has the method get_price() which returns price, and set_price() to ensure price is numeric.

Then create two instances of this class:

b1:
* title -> 'Practical Programming'
* author -> 'Gries, Campbell, Montojo'
* pages -> 383
* price -> 50

b2:
* title -> 'Building a Career in Data Science'
* author -> 'Robinson, Nolis'
* pages -> 322
* price -> 40

```python
# Your code goes here
class Book:
    def __init__(self, title, author, pages, price):
        self.title = title
        self.author = author
        self.pages = pages
        self.price = price

    def get_price(self):
        return self.price

    def set_price(self, price):
        self.price = price

book_1 = Book('Practical Programming', 'Gries, Campbell, Montojo', 383, 50)
book_2 = Book('Building a Career in Data Science', 'Robinson, Nolis', 322, 40)
```

```python
print(f"Price of '{book_1.title}': ${book_1.get_price()}")
print(f"Price of '{book_2.title}': ${book_2.get_price()}")
```
